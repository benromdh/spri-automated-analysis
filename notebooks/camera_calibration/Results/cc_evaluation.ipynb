```python
from analysis.camera_calibration import * 
from segmentation.segmenter import *
from PIL import Image
import pandas as pd
import os
from tqdm.notebook import tqdm

def setup_camera_calibration(setup_image: np.ndarray, camera: "Camera", debug: bool = False) -> "CameraCalibrator":
    # Get reference mask
    model = CellposeSAMWrapper()
    segmenter = Segmenter(model=model)
    segmenter.generate_masks(setup_image)
    segmenter.filter_masks()
    setup_mask = segmenter.large_masks.masks[0]
    
    if debug:
        segmenter.visualize_masks_by_well_size()
    
    calibrator = CameraCalibrator(
        setup_image=setup_image,
        setup_mask=setup_mask,
        camera=CAMERA
    )
    
    # do calculation already to speed up runtime
    calibrated = calibrator.undistort(setup_image)

    if debug:
        _, ax = plt.subplots(1, 2, figsize=(12, 6))
        ax[0].imshow(setup_image, cmap='gray')
        ax[0].set_title("Raw Image")
        ax[0].axis('off')
        
        ax[1].imshow(calibrated, cmap='gray')
        ax[1].set_title("Calibrated Image")
        ax[1].axis('off')
        plt.tight_layout()
        plt.show()

    return calibrator, calibrated

cellposeSAM = CellposeSAMWrapper()
segmenter = Segmenter(model=cellposeSAM)

image_path = "../../data/Raw/Chip2-NaCl 0-19_/"
image_raw = np.array(Image.open(image_path+"290-1-0,75x-0_-1.jpg").convert('L'))
    
CAMERA = Camera(
    zoom=0.75,
    um_per_pixel=6.45
)
calibrator, calibrated_image = setup_camera_calibration(image_raw, CAMERA, debug=False)
mc = segmenter.generate_masks(calibrated_image)
resize = calculate_resize(mc)

schematic_path = "../../data/Raw/Chip-schematic/chip_cleaned_1_10.jpg"

schematic = np.array(Image.open(schematic_path).convert('L'))
schematic_mask = extract_schematic(image=schematic)

df = pd.DataFrame(columns=["iou_all", "iou_large", "iou_medium", "iou_small"])




def split_into_individual_masks(binary_image):
    # Ensure binary format (0 and 255), and uint8 type
    if binary_image.dtype == bool:
        binary_image = binary_image.astype(np.uint8) * 255
    elif binary_image.dtype != np.uint8:
        binary_image = (binary_image > 0).astype(np.uint8) * 255

    # Label connected components
    num_labels, labels = cv2.connectedComponents(binary_image, connectivity=8)

    # List to hold individual binary masks
    individual_masks = []

    for label in range(1, num_labels):  # skip background (label 0)
        mask = (labels == label).astype(np.uint8) * 255
        individual_masks.append(mask)

    return individual_masks


def find_overlapping_pairs(mc: MaskContainer, gt_mc: MaskContainer, min_overlap_pixels: int = 5) -> [list, list]:
    src_pts = []
    dst_pts = []
    for mask in mc.masks:
        for gt_mask in gt_mc.masks:
            overlap_pixels = np.sum(mask.segmentation & gt_mask.segmentation)
            if overlap_pixels > min_overlap_pixels:
                src_pts.append(mask.center)
                dst_pts.append(gt_mask.center)
                break
    return np.array(src_pts), np.array(dst_pts)

 

for file in tqdm([f for f in os.listdir(image_path) if f.split('.')[0].endswith('1')]):
    name = file.split('.')[0]
    if name.endswith('1'):
        print(f"==== Evaluating camera calibration for image {name}. ====")
        image_raw = np.array(Image.open(os.path.join(image_path, file)).convert("L"))
        affine_image = calibrator.undistort(image_raw)
        aligned_schematic, iou_all, iou_large, iou_medium, iou_small = evaluate_camera_calibration(affine_image, schematic_mask, segmenter, resize=resize, debug=True)
        if aligned_schematic is None:
            continue # if no masks detected the alignment fails
        df.loc[len(df)] = [iou_all, iou_large, iou_medium, iou_small]
        
        ### Use perspective correction between centroids and schematic centroids
        ## Step 1: Create corresponding centroid pairs
        # Resize schematic, split into individual masks and label their centroids 
        # resized_schematic, _ = resize_uniform(schematic_mask, resize)
        # padded_schematic = pad_to_shape(resized_schematic, affine_image.shape)
        
        # individual_masks = split_into_individual_masks(aligned_schematic)
        # gt_mc = MaskContainer(
        #     image=schematic_mask,
        #     masks=[WellMask(segmentation=mask) for mask in individual_masks]
        # )

        # # Segment affine corrected image
        # segmenter.generate_masks(affine_image)
        # segmenter.filter_masks()
        # segmenter.visualize_masks()
        # segmenter.set_image(aligned_schematic.astype(np.uint8)*255)
        # segmenter.visualize_masks(mc=gt_mc)
        # mc = segmenter.small_masks
        # segmenter.visualize_masks(mc)
        
        # # Find overlapping pairs and create corresponding centroid lists
        # src_pts, dst_pts = find_overlapping_pairs(mc, gt_mc)
        # print(f"Src: {len(src_pts)}, Dst: {len(dst_pts)}")
        # for i in range(len(src_pts)):
        #     print(f"Src: {src_pts[i]} --> Dst: {dst_pts[i]}")

        # ## Step 2: Use cv2.findHomography
        # H, status = cv2.findHomography(src_pts, dst_pts, method=cv2.RANSAC)

        # ## Step 3: Apply transformation
        # corrected_img = cv2.warpPerspective(affine_image, H, (affine_image.shape[1], affine_image.shape[0]))

        # plt.figure(figsize=(10,10))
        # plt.imshow(corrected_img, cmap='gray')
        # plt.show()
        
        # _, iou_all, iou_large, iou_medium, iou_small = evaluate_camera_calibration(corrected_img, schematic_mask, segmenter, resize=resize, debug=True)
        # df.loc[len(df)] = [iou_all, iou_large, iou_medium, iou_small]
```
